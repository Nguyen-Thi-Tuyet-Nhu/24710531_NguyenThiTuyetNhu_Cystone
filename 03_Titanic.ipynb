## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO
        # Tỷ lệ sống sót theo hạng vé
survival_rate = df.groupby('class')['survived'].mean() * 100


        # Chênh lệch cao nhất và thấp nhất
difference = survival_rate.max() - survival_rate.min()



print("Hạng vé sống sót cao nhất:", survival_rate.idxmax())
print("Tỷ lệ sống sót:", round(survival_rate.max(), 2), "%")
print("Chênh lệch:", round(difference, 2), "%")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
# Tỷ lệ sống sót theo giới tính
gender_survival = df.groupby('sex')['survived'].mean() * 100
print("Tỷ lệ sống sót theo giới tính:")
print(gender_survival)


# Tỷ lệ sống sót theo hạng vé
class_survival = df.groupby('class')['survived'].mean() * 100
print("\nTỷ lệ sống sót theo hạng vé:")
print(class_survival)
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
# Chia giá vé thành 3 nhóm
df['fare_group'] = pd.qcut(df['fare'], 3)

# Tính tỷ lệ sống sót theo giá vé
fare_survival = df.groupby('fare_group', observed=True)['survived'].mean() * 100

print(fare_survival)
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
# Tính số người trong gia đình
df['family_size'] = df['sibsp'] + df['parch'] + 1

# Tính tỷ lệ sống sót theo quy mô gia đình
family_survival = df.groupby('family_size')['survived'].mean() * 100

print(family_survival)
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
# TODO***********************************
# Tỷ lệ sống sót theo cảng lên tàu
embark_survival = df.groupby('embark_town')['survived'].mean() * 100

print(embark_survival)

# Cảng có tỷ lệ sống sót cao nhất
print("Cảng có tỷ lệ sống sót cao nhất:", embark_survival.idxmax())
print("Tỷ lệ:", round(embark_survival.max(), 2), "%")
```
